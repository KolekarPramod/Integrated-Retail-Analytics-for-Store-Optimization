# What Do MarkDown1 to MarkDown5 Represent?


In the retail industry, a markdown refers to a permanent reduction in the original selling price of a product. This is done to increase the product’s attractiveness to customers and stimulate sales, especially when:
 The product is not selling well at its original price.
 Seasonal items need to be cleared out.
 The store wants to make room for new inventory.
 The item is damaged or obsolete.

These likely represent different types or phases of markdowns applied to products in a retail setting. Here's a likely interpretation based on standard retail practices:
ColumnMeaning (Likely)

1. MarkDown1: Initial or minor discount – e.g., 10–20% off
2. MarkDown2: Follow-up reduction for low-performing items
3. MarkDown3: Further markdown – more aggressive, e.g., 30%+
4. MarkDown4: Clearance-level markdowns
5. MarkDown5: Final or deep clearance discount (e.g., 50%+)



###Why Are There Many Null Values?
 Not all products go through all markdown phases.
 Some weeks or stores may not apply certain markdowns.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
features_df['Date'] = pd.to_datetime(features_df['Date'],dayfirst=True)
sales_df['Date'] = pd.to_datetime(sales_df['Date'],dayfirst=True)
````

````python
features_df.info()
````

# Merging the Data

````python
# Merge different datasets
df = pd.merge(sales_df, features_df, how = 'left', on = ['Store','Date','IsHoliday']) # merging sales and features by left since there are more rows for sales
final_df = pd.merge(df, stores_df, how = 'left', on = 'Store') #merge df and df_stores by left to keep all the columns and rows
````

````python
#Check for missing values
final_df.isnull().sum()
````

````python
# Set your DataFrame name
markdown_cols = ['MarkDown1', 'MarkDown2', 'MarkDown3', 'MarkDown4', 'MarkDown5']

plt.figure(figsize=(20, 10))

for i, col in enumerate(markdown_cols, 1):
    plt.subplot(2, 3, i)
    # Histogram ignoring NaNs
    sns.histplot(final_df[col].dropna(), bins=50, kde=True, color='skyblue')
    plt.title(f'Histogram of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')

    # Compute and display mean/median/skew
    mean_val = final_df[col].mean()
    median_val = final_df[col].median()
    skew_val = final_df[col].skew()

    plt.axvline(mean_val, color='red', linestyle='dashed', linewidth=1, label=f"Mean: {mean_val:.2f}")
    plt.axvline(median_val, color='green', linestyle='dashed', linewidth=1, label=f"Median: {median_val:.2f}")
    plt.legend()

plt.tight_layout()
plt.show()

````

````python
# Console output for skewness and null count
for col in markdown_cols:
    mean_val = final_df[col].mean()
    median_val = final_df[col].median()
    skew_val = final_df[col].skew()
    null_count = final_df[col].isna().sum()
    print(f"{col}: Skewness={skew_val:.3f} | Mean={mean_val:.2f} | Median={median_val:.2f} | Nulls={null_count}")
````

````python
final_df = final_df.fillna(0)
````

````python
# Console output for skewness and null count
for col in markdown_cols:
    mean_val = final_df[col].mean()
    median_val = final_df[col].median()
    skew_val = final_df[col].skew()
    null_count = final_df[col].isna().sum()
    print(f"{col}: Skewness={skew_val:.3f} | Mean={mean_val:.2f} | Median={median_val:.2f} | Nulls={null_count}")
````
